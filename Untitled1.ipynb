```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import warnings
warnings.filterwarnings('ignore')

print("TensorFlow version:", tf.__version__)
```

```python
df = pd.read_csv('customer_churn_nn.csv')

print("=" * 50)
print("TASK 1: DATASET UNDERSTANDING")
print("=" * 50)

print(f"\n📌 Shape: {df.shape[0]} rows, {df.shape[1]} columns")

print("\n📌 Column Data Types:")
print(df.dtypes)

print("\n📌 Target Variable (churn) Distribution:")
print(df['churn'].value_counts())
print(f"\n  Churned: {df['churn'].sum()} ({df['churn'].mean()*100:.1f}%)")
print(f"  Retained: {(df['churn']==0).sum()} ({(df['churn']==0).mean()*100:.1f}%)")

print("\n📌 Missing Values:")
print(df.isnull().sum())
print(f"Total missing: {df.isnull().sum().sum()}")

print("\n📌 Basic Statistical Summary:")
print(df.describe())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Churn count
axes[0].bar(['Retained (0)', 'Churned (1)'], df['churn'].value_counts().values,
            color=['steelblue', 'tomato'])
axes[0].set_title('Target Variable Distribution')
axes[0].set_ylabel('Count')
for i, v in enumerate(df['churn'].value_counts().values):
    axes[0].text(i, v + 5, str(v), ha='center', fontweight='bold')

# Satisfaction score by churn
axes[1].boxplot([df[df['churn']==0]['satisfaction_score'],
                 df[df['churn']==1]['satisfaction_score']],
                labels=['Retained', 'Churned'])
axes[1].set_title('Satisfaction Score by Churn')
axes[1].set_ylabel('Satisfaction Score')

plt.tight_layout()
plt.savefig('evaluation_outputs.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved as evaluation_outputs.png")
```

```python
print("=" * 50)
print("TASK 2: DATA PREPROCESSING")
print("=" * 50)

# Drop customer_id (identifier, not a feature)
df_clean = df.drop(columns=['customer_id'])

# Categorical columns to encode
cat_cols = ['region', 'plan_type', 'contract_type', 'payment_method']
num_cols = ['tenure_months', 'monthly_charges_inr', 'avg_login_days_per_month',
            'support_tickets_last_90_days', 'payment_delay_days', 'data_usage_gb',
            'satisfaction_score', 'last_complaint_days_ago', 'discount_percent',
            'autopay_enabled', 'referral_count']

# No missing values, confirmed above. Encode categoricals with get_dummies
df_encoded = pd.get_dummies(df_clean, columns=cat_cols, drop_first=True)
print(f"Shape after encoding: {df_encoded.shape}")

# Separate features and target
X = df_encoded.drop(columns=['churn'])
y = df_encoded['churn']

# Scale numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train/test split (80/20)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y)

print(f"\n✅ Training set: {X_train.shape}")
print(f"✅ Testing set:  {X_test.shape}")
print(f"✅ Input features: {X.shape[1]}")
print(f"\nEncoded columns: {list(X.columns)}")
```

```python
print("=" * 50)
print("TASK 3: NEURAL NETWORK MODEL BUILDING")
print("=" * 50)

input_dim = X_train.shape[1]

def build_model(hidden_layers=1, neurons=32, activation='relu', learning_rate=0.001):
    model = keras.Sequential()
    # Input layer
    model.add(layers.Input(shape=(input_dim,)))
    # Hidden layers
    for _ in range(hidden_layers):
        model.add(layers.Dense(neurons, activation=activation))
    # Output layer (binary classification → sigmoid)
    model.add(layers.Dense(1, activation='sigmoid'))

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model

# Default model
model = build_model(hidden_layers=1, neurons=32, activation='relu', learning_rate=0.001)
model.summary()
```

```python
print("=" * 50)
print("TASK 4: TRAINING AND EVALUATION")
print("=" * 50)

# Handle class imbalance with class weights
from sklearn.utils.class_weight import compute_class_weight
class_weights = compute_class_weight('balanced', classes=np.array([0, 1]), y=y_train)
class_weight_dict = {0: class_weights[0], 1: class_weights[1]}
print(f"Class weights: {class_weight_dict}")

# Train
history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=32,
    validation_data=(X_test, y_test),
    class_weight=class_weight_dict,
    verbose=1
)
```

```python
# Evaluate on test set
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
train_loss, train_acc = model.evaluate(X_train, y_train, verbose=0)

print(f"\n📊 Training Accuracy:  {train_acc:.4f}  | Training Loss:  {train_loss:.4f}")
print(f"📊 Testing  Accuracy:  {test_acc:.4f}  | Testing  Loss:  {test_loss:.4f}")

# Confusion Matrix
y_pred = (model.predict(X_test) > 0.5).astype(int).flatten()

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Loss/Accuracy curves
axes[0].plot(history.history['accuracy'], label='Train Accuracy')
axes[0].plot(history.history['val_accuracy'], label='Val Accuracy')
axes[0].set_title('Accuracy over Epochs')
axes[0].set_xlabel('Epoch')
axes[0].legend()

axes[1].plot(history.history['loss'], label='Train Loss')
axes[1].plot(history.history['val_loss'], label='Val Loss')
axes[1].set_title('Loss over Epochs')
axes[1].set_xlabel('Epoch')
axes[1].legend()

plt.tight_layout()
plt.savefig('evaluation_outputs.png', dpi=150, bbox_inches='tight')
plt.show()

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Retained', 'Churned'])
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - Default Model')
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150)
plt.show()

print("\n📋 Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Retained', 'Churned']))
```

```python
print("=" * 50)
print("TASK 5: HYPERPARAMETER EXPERIMENTATION")
print("=" * 50)

experiments = [
    {"name": "Baseline",          "hidden_layers": 1, "neurons": 32,  "lr": 0.001, "batch": 32,  "epochs": 50, "activation": "relu"},
    {"name": "More Neurons",      "hidden_layers": 1, "neurons": 64,  "lr": 0.001, "batch": 32,  "epochs": 50, "activation": "relu"},
    {"name": "2 Hidden Layers",   "hidden_layers": 2, "neurons": 32,  "lr": 0.001, "batch": 32,  "epochs": 50, "activation": "relu"},
    {"name": "High LR (0.01)",    "hidden_layers": 1, "neurons": 32,  "lr": 0.01,  "batch": 32,  "epochs": 50, "activation": "relu"},
    {"name": "Low LR (0.0001)",   "hidden_layers": 1, "neurons": 32,  "lr": 0.0001,"batch": 32,  "epochs": 50, "activation": "relu"},
    {"name": "Larger Batch (64)", "hidden_layers": 1, "neurons": 32,  "lr": 0.001, "batch": 64,  "epochs": 50, "activation": "relu"},
    {"name": "tanh Activation",   "hidden_layers": 1, "neurons": 32,  "lr": 0.001, "batch": 32,  "epochs": 50, "activation": "tanh"},
]

results = []
for exp in experiments:
    print(f"\n🔄 Running: {exp['name']} ...")
    m = build_model(
        hidden_layers=exp['hidden_layers'],
        neurons=exp['neurons'],
        activation=exp['activation'],
        learning_rate=exp['lr']
    )
    m.fit(X_train, y_train,
          epochs=exp['epochs'],
          batch_size=exp['batch'],
          class_weight=class_weight_dict,
          verbose=0)

    tr_loss, tr_acc = m.evaluate(X_train, y_train, verbose=0)
    te_loss, te_acc = m.evaluate(X_test,  y_test,  verbose=0)

    yp = (m.predict(X_test, verbose=0) > 0.5).astype(int).flatten()
    from sklearn.metrics import f1_score
    f1 = f1_score(y_test, yp, zero_division=0)

    results.append({
        "Experiment":      exp['name'],
        "Hidden Layers":   exp['hidden_layers'],
        "Neurons":         exp['neurons'],
        "Learning Rate":   exp['lr'],
        "Batch Size":      exp['batch'],
        "Activation":      exp['activation'],
        "Train Acc":       round(tr_acc, 4),
        "Test Acc":        round(te_acc, 4),
        "Test Loss":       round(te_loss, 4),
        "F1 Score":        round(f1, 4),
    })
    print(f"   Train Acc: {tr_acc:.4f} | Test Acc: {te_acc:.4f} | F1: {f1:.4f}")

results_df = pd.DataFrame(results)
print("\n\n📊 Comparison Table:")
print(results_df.to_string(index=False))
```

```python
# Save as CSV
results_df.to_csv('model_comparison_table.csv', index=False)
print("✅ model_comparison_table.csv saved")

# Visualize as a styled table image
fig, ax = plt.subplots(figsize=(16, 3))
ax.axis('off')
table = ax.table(
    cellText=results_df.values,
    colLabels=results_df.columns,
    cellLoc='center',
    loc='center'
)
table.auto_set_font_size(False)
table.set_fontsize(8)
table.scale(1, 1.6)
plt.title('Hyperparameter Experiment Results', fontsize=13, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('model_comparison_table.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ model_comparison_table.png saved")
```

## Task 6: Final Reflection

### 1. Role of Weights and Biases
Weights control how strongly each input feature influences the output —
they are the learnable parameters the network adjusts during training.
Biases allow the model to shift the activation function, helping it fit
data that doesn't pass through the origin. Together, they define the
decision boundary the network learns.

### 2. Why Activation Functions Are Required
Without activation functions, stacking multiple layers would still
produce a purely linear model (composition of linear functions = linear).
Activation functions like ReLU or tanh introduce non-linearity, allowing
the network to learn complex patterns in the data.

### 3. Learning Rate: Too High vs Too Low
- **Too high**: The optimizer overshoots the minimum — loss oscillates
  or diverges instead of converging.
- **Too low**: Training is extremely slow, and the model may get stuck
  in a local minimum and never reach good performance within a reasonable
  number of epochs.

### 4. Underfitting vs Overfitting
Given the severe class imbalance (only ~1.55% churn rate), the model
initially tended toward predicting "Retained" for everything (high
accuracy, low recall). Using class weights helped the model detect
churners better. Validation loss tracking and the F1 score per
experiment reveal whether the model overfits (val loss rising while
train loss falls) or underfits (both losses remain high).
